First iteration of selected features to keep

```python
columns = [
    # financial indicators 
    "NETWORTH",      
    "INCOME",
    "FIN",
    "NFIN",
    "DEBT",

    # investments / assets
    "EQUITY",
    "STOCKS",
    "HHOUSES",
    "BUS",

    # retirement-specific
    "RETQLIQ",

    # behavior / risk
    "YESFINRISK",
    "SPENDMOR",
    "LATE",
    "EMERGSAV",

    # liquidity
    "LIQ",

    # financial capability
    "FINLIT",

    # demographics
    "AGE",
    "EDUC",
    "MARRIED",
    "KIDS",

    # employment proxy
    "OCCAT1",
]
```

```python
filtered_df = df_si[columns]
filtered_df.shape
```

```python
filtered_df['RETQLIQ']
```

Derived RET_ASSETS combines broader financial assets and retirement liquid assets to better capture households with substantial non-retirement financial resources.

```python
# create improved retirement asset target variable
filtered_df["RET_ASSETS"] = filtered_df["FIN"] + filtered_df["RETQLIQ"]

filtered_df[["FIN", "RETQLIQ", "RET_ASSETS"]].head(20)
```

```python
percent_fin_greater = (filtered_df["FIN"] > filtered_df["RETQLIQ"]).mean() * 100

print(f"{percent_fin_greater:.2f}% of rows have FIN greater than RETQLIQ")
```
